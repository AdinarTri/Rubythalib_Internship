### Image Cropping

[Documentation](https://opencv.org/blog/cropping-an-image-using-opencv/)

```python
from google.colab import files
from IPython.display import Image
import cv2
import numpy as np

uploaded = files.upload()

for fn in uploaded.keys():
  print('User uploaded file "{name}" with length {length} bytes'.format(
      name=fn, length=len(uploaded[fn])))
  display(Image(fn))

  # Read the uploaded image using OpenCV
  image = cv2.imread(fn)

  # Define coordinates for cropping (y_start:y_end, x_start:x_end)
  # Replace these values with your desired coordinates
  y_start = 50
  y_end = 200
  x_start = 100
  x_end = 300

  cropped_image = image[y_start:y_end, x_start:x_end]

  # Save the cropped image
  cv2.imwrite('cropped_image.jpg', cropped_image)

  print("Image cropped and saved as 'cropped_image.jpg'")
```

```python
import cv2

# Baca gambar
image = cv2.imread('/content/wallpapersden.com_one-piece-egghead_1920x1350.jpg')

# Tentukan koordinat untuk cropping (x, y, width, height)
cropped_image = image[50:200, 100:300]


# Simpan hasil cropping
cv2.imwrite('cropped_image.jpg', cropped_image)
```

### Line Detection

```python
from google.colab import files
from IPython.display import Image
import cv2
import numpy as np

uploaded = files.upload()

for fn in uploaded.keys():
  print('User uploaded file "{name}" with length {length} bytes'.format(
      name=fn, length=len(uploaded[fn])))
  display(Image(fn))

  # Read the uploaded image using OpenCV
  image = cv2.imread(fn)

  gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

  # Deteksi tepi menggunakan Canny
  edges = cv2.Canny(gray, 50, 150, apertureSize=3)

  # Deteksi garis menggunakan Hough Transform
  lines = cv2.HoughLines(edges, 1, np.pi / 180, 200)

  # Gambar garis pada gambar asli

  if lines is not None:
    for rho, theta in lines[:, 0]:
      a = np.cos(theta)
      b = np.sin(theta)
      x0 = a * rho
      y0 = b * rho
      x1 = int(x0 + 1000 * (-b))
      y1 = int(y0 + 1000 * (a))
      x2 = int(x0 - 1000 * (-b))
      y2 = int(y0 - 1000 * (a))
      cv2.line(image, (x1, y1), (x2, y2), (0, 0, 255), 2)

  # Simpan hasil deteksi garis
  cv2.imwrite('line_detected_image.jpg', image)
```

```python
import cv2
import numpy as np

# Baca gambar dan konversi ke grayscale
image = cv2.imread('input_image.png')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Deteksi tepi menggunakan Canny
edges = cv2.Canny(gray, 50, 150, apertureSize=3)

# Deteksi garis menggunakan Hough Transform
lines = cv2.HoughLines(edges, 1, np.pi / 180, 200)

# Gambar garis pada gambar asli

if lines is not None:
 for rho, theta in lines[:, 0]:
  a = np.cos(theta)
  b = np.sin(theta)
  x0 = a * rho
  y0 = b * rho
  x1 = int(x0 + 1000 * (-b))
  y1 = int(y0 + 1000 * (a))
  x2 = int(x0 - 1000 * (-b))
  y2 = int(y0 - 1000 * (a))
  cv2.line(image, (x1, y1), (x2, y2), (0, 0, 255), 2)


# Simpan hasil deteksi garis
cv2.imwrite('line_detected_image.jpg', image)
```

Edge Detection [Documentation](https://opencv.org/blog/edge-detection-using-opencv/)

```python
import numpy as np
import cv2 as cv
from matplotlib import pyplot as plt

img = cv.imread('Image_1.png', cv.IMREAD_GRAYSCALE)
assert img is not None, "file could not be read, check with os.path.exists()"
edges = cv.Canny(img,100,200)

plt.subplot(121),plt.imshow(img,cmap = 'gray')
plt.title('Original Image'), plt.xticks([]), plt.yticks([])
plt.subplot(122),plt.imshow(edges,cmap = 'gray')
plt.title('Edge Image'), plt.xticks([]), plt.yticks([])

plt.show()
```

```python
# Apply Sobel operator
sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)  # Horizontal edges
sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)  # Vertical edges

# Compute gradient magnitude
gradient_magnitude = cv2.magnitude(sobelx, sobely)

# Convert to uint8
gradient_magnitude = cv2.convertScaleAbs(gradient_magnitude)

cv2.imwrite('sobel_edge_detection.jpg', gradient_magnitude)
```

```python
# Apply Laplacian operator
laplacian = cv2.Laplacian(img, cv2.CV_64F)

# Convert to uint8
laplacian_abs = cv2.convertScaleAbs(laplacian)

cv2.imwrite('laplacian_detection.jpg', laplacian_abs)
```

```python
# Apply Gaussian Blur to reduce noise
blur = cv2.GaussianBlur(img, (5, 5), 1.4)

# Apply Canny Edge Detector
edges = cv2.Canny(blur, threshold1=100, threshold2=200)

cv2.imwrite('canny_edge_detection.jpg', edges)
```
